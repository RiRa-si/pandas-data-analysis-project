```python
# Q2 — Conditional Filtering
# Find employees with Salary > 50000 and Performance > 85.
# Display Name, Salary, and Performance.

employees[
    employees["Department"].isin(["IT", "Sales"])
][["Name", "Department", "Salary"]]
```

```python
# Q4 — Using .between()
# Find employees whose Age is between 25 and 32 (inclusive).
# Display Name, Age, and Department.


employees[
    employees["Age"].between(25, 32)
][["Name", "Age", "Department"]]
```

```python
# Q5 — Multiple Conditions
# Find employees who:
# - Belong to IT or Sales
# - Have Salary between 40000 and 60000
# - Have Performance > 80
# Display Name, Department, Salary, and Performance.
employees[
    (employees["Department"].isin(["IT", "Sales"])) &
    (employees["Salary"].between(40000, 60000)) &
    (employees["Performance"] > 80)
][["Name", "Department", "Salary", "Performance"]]
```

```python
# Q6 — Create a New Column
# Create Annual_Salary = Salary * 12.

employees["Annual_Salary"] = employees["Salary"] * 12
employees["Annual_Salary"]
```

```python
# Q7 — Performance Level
# Create Performance_Level using np.where():
# - Performance >= 90 → Excellent
# - Performance >= 80 → Good
# - Performance >= 70 → Average
# - Otherwise → Needs Improvement

employees["Performance_Level"] = np.where(
    employees["Performance"] >= 90,
    "Excellent",
    np.where(
        employees["Performance"] >= 80,
        "Good",
        np.where(
            employees["Performance"] >= 70,
            "Average",
            "Needs Improvement"
        )
    )
)

employees["Performance_Level"]
```

```python
# Q8 — Extract First Name
# Create First_Name by extracting the first name from Name.
# Example: "Aman Kumar" → "Aman"

employees["First_Name"] = employees["Name"].str.split(" ").str[0]
employees["First_Name"]
```

```python
# Q9 — Extract and Filter Last Name
# Extract the last name from Name.
# Find employees whose last name is Kumar.
# Display Name, Department, and Salary.

employees["Last_name"] = employees["Name"].str.split(" ").str[1]

employees[
    employees["Last_name"] == "Kumar"
][["Name", "Department", "Salary"]]
```

```python
# Q10 — String Transformation
# Create Name_Upper containing names in uppercase.

employees["Name_Upper"] = employees["Name"].str.upper()
employees["Name_Upper"]
```

```python
#Q18: high performers 
#find employees qwho satisfy all three:
employees[
    (employees["Performance"] >= 85) & 
    (employees["Salary"] > 40000)&
    (employees["Experience"]>=3)
    ][["Name","Department","Salary","Experience","Performance"]].sort_values(by = "Performance",ascending = False)
```

```python
# Q20 — Handle Missing Age
# Fill missing Age values using the median Age.
# Verify that no missing values remain in Age.
employees["Age"] = employees["Age"].fillna(employees["Age"].median())
employees["Age"].isna().sum()
```

```python
# Q21 — Duplicate Check
# Check for duplicate rows.
# Display duplicate rows and count the total number of duplicates.
employees[employees.duplicated()]
employees.duplicated().sum()
```

```python
#Dataset 2
departments = pd.DataFrame({
    "Department": ["Sales", "IT", "HR"],
    "Manager": ["Mr. Verma", "Ms. Sharma", "Mr. Singh"],
    "Office": [101, 202, 303]
})
```

```python
# Q22 — Merge / SQL JOIN
# Merge employees and departments using Department.
# Use an INNER JOIN.
# Display Name, Department, Manager, and Office.
pd.merge(
    employees,
    departments,
    on ="Department",
    how = "inner"
)[["Name","Department","Manager","Office"]]
```

```python
#23: extract manger's last nam
# Q23 — Extract Manager's Last Name
# Create Manager_Last_Name from Manager.
# Example: "Mr. Verma" → "Verma"
departments["Mangers_last_name"]=departments["Manager"].str.split(" ").str[-1]
```

```python
# Q24 — Filter After Merge
# From the merged DataFrame, find employees who:
# - Belong to IT or Sales
# - Have Performance > 85
# Display Name, Department, Manager, and Performance.
merged =pd.merge(
    employees,
    departments,
    on = "Department",
    how = "inner"
)
merged[
    merged["Department"].isin(["IT","Sales"]) &
    (merged["Performance"]>85)
    ][["Name","Department","Manager","Performance"]]
```

```python
# Q25 — Complete Pandas Analysis
# 1. Create Annual_Salary.
# 2. Create Performance_Level.
# 3. Create First_Name.
# 4. Merge employees with departments.
# 5. Keep employees who:
#    - Belong to IT or Sales
#    - Have Performance >= 80
#    - Have Experience >= 3
# 6. Display:
#    - First_Name
#    - Department
#    - Manager
#    - Annual_Salary
#    - Experience
#    - Performance
#    - Performance_Level
# 7. Sort by Performance descending.

#annual salary:
employees["Annual_Salary"] = employees["Salary"] * 12
employees["Annual_Salary"]
#first name :
employees["First_Name"] = employees["Name"].str.split(" ").str[0]
employees["First_Name"]
#performance level
employees["Performance_Level"] = np.where(
    employees["Performance"] >= 90,
    "Excellent",
    np.where(
        employees["Performance"] >= 80,
        "Good",
        np.where(
            employees["Performance"] >= 70,
            "Average",
            "Needs Improvement"
        )
    )
)

employees["Performance_Level"]

#merged 2 
merged2 =pd.merge(
    employees,
    departments,
    on = "Department",
    how = "inner"
)

keep_emp = merged2[merged2["Department"].isin(["IT","Sales"]) &
           (merged2["Performance"]>=80) &
           (merged2["Experience"]>=3)]

# Select the requested columns and sort by Performance in descending order
result = keep_emp[[
    "First_Name", 
    "Department", 
    "Manager", 
    "Annual_Salary", 
    "Experience", 
    "Performance", 
    "Performance_Level"
]].sort_values(by="Performance", ascending=False)

# Display the resulting DataFrame
print(result)
```

```python
# Q26 — Final Department Analytics
# Create a department-level report containing:
# - Department
# - Employees
# - Avg_Salary
# - Avg_Experience
# - Avg_Performance
# - Max_Salary
# Sort by Avg_Performance descending.

#Final Department Analytics
result2 = employees.groupby("Department").agg(
    Employees=("EmployeeID", "count"),
    Avg_Salary=("Salary", "mean"),
    Avg_Experience=("Experience", "mean"),
    Avg_Performance=("Performance", "mean"),
    Max_Salary=("Salary", "max")
).reset_index()

result3= result2.sort_values(
    by="Avg_Performance",
    ascending = False
)
result3
```
